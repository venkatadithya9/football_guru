## Leagues

Basically gives context on what data is being tracked for a given league, and across how many seasons. 
The ids are given at this [website](https://dashboard.api-football.com/soccer/ids). 
- Premier League : 39
- LaLiga : 140
- Bundesliga: 79
- Seria A : 135
- Ligue 1: 61


```python
response = requests.get(os.path.join(url, "leagues?id=39"), headers=headers)
if response.status_code == 200:
    if 'response' in response.json():
        leagues_data = response.json()['response']
        print(leagues_data.keys())
    pprint(leagues_data)
```

```python
print(leagues_data[0].keys())
print(leagues_data[0]['league'].keys())
print(leagues_data[0]['country'].keys())
print(leagues_data[0]['seasons'][0].keys())
print(leagues_data[0]['seasons'][0]['coverage'].keys())
print(leagues_data[0]['seasons'][0]['coverage']['standings'])
```
