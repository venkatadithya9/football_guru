## Leagues

Basically gives context on what data is being tracked for a given league, and across how many seasons. 
The ids are given at this [website](https://dashboard.api-football.com/soccer/ids). 
- Premier League : 39
- LaLiga : 140
- Bundesliga: 79

```python
response = requests.get(os.path.join(url, "leagues?id=39"), headers=headers)
if response.status_code == 200:
    if 'response' in response.json():
        
        leagues_data = response.json()['response']
    pprint(leagues_data)
```
