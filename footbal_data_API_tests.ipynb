```python
import pandas as pd
import numpy as np
import requests
import os
from rich import print as pprint
```

```python
url = "http://api.football-data.org/v4/"
```

## Available subtopics in the API
1. areas:
    - areas/ : List all available areas (countries)
    - areas/{id}: List one particular area (country)
2. comptetions
    - competitions/ : List all competitions that are being tracked
3. teams
4. persons
5. matches

## Areas

```python
## Areas
# List areas /areas
response = requests.get(os.path.join(url, "areas"))
if response.status_code == 200:
    data = response.json()
    pprint(data)
```

```python
# Get specific area
id = "2072"
response = requests.get(os.path.join(url, "areas", id))
if response.status_code == 200:
    data = response.json()
    pprint(data)
    
```

## Competitions

```python
# List competitions
response = requests.get(os.path.join(url, "competitions"))
if response.status_code == 200:
    data = response.json()
    pprint(data)
```
